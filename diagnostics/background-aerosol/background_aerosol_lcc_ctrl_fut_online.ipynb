# Background aerosol, model response and online sources: LCC_FUT vs CTRL_FUT

**Question.** Does the forest expansion change the aerosol and chemistry background of the FUT runs, and through which sources?

**Runs.** CTRL_FUT `NF2100ssp585norbc_tropstratchem_nudg_ctrl_f19_f19-20261028` and LCC_FUT `NF2100ssp585norbc_tropstratchem_nudg_lcc_f19_f19-20261028` (new chain), **2001–2025** (first 12 months dropped as spin-up). Both are nudged to the same PD winds, so year *y* of one run is paired with year *y* of the other.

**What can differ.** The two runs share every prescribed input (§1). The land cover differs at 46–82°N, so differences come from the land surface, the land-dependent online sources (**MEGAN BVOC**, **dust**) and everything downstream. Sea salt and ocean DMS can change only through the 10-m wind.

**Twin notebook:** `background_aerosol_pd_fut_online.ipynb` runs **the same analysis** on CTRL_FUT vs CTRL_PD. Prescribed inputs: `background_aerosol_pd_fut_prescribed.ipynb`.

**Same structure in both notebooks.**
1. Inputs: namelist differences between the two runs.
2. Online sources: MEGAN BVOC, dust, sea salt, ocean DMS flux, lightning NOx: regional tables, maps, totals and time series, global and north of 40°N.
3. Model response: AOD, absorption AOD, SOA, sulfate, droplets, surface BVOC, OH, SO₂, DMS, NOx, O₃, CO, CH₄, O₃ column.
4. Summary. Appendix: every map, global.

**Statistics.** Yearly means; paired t-test of the 25 yearly differences (`analysis/stats.py` method, copied here). Maps: Benjamini–Hochberg FDR (α = 0.10) over the plotted domain; **hatching = not significant**. Years are treated as independent, and table p-values are not corrected for the many variables and regions.

**Regions.** global, NH, SH, **46–82°N** (the only latitudes where the LCC land cover differs), **modified cells** (land cells whose PFTs changed, `diagnostics/lcc-modified-cells/modified_cells.nc`), Arctic >66.5°N. Maps: Arctic (north of 45°N, dashed = 66.5°N, black line = modified cells); global versions in the Appendix (`figures/appendix-global/`).

```python
import os
os.environ.setdefault('HDF5_USE_FILE_LOCKING', 'FALSE')  # before netCDF4/xarray import
import re, difflib, warnings
from pathlib import Path
from concurrent.futures import ProcessPoolExecutor
import numpy as np
import pandas as pd
import xarray as xr
import netCDF4
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
import matplotlib.path as mpath
import cartopy.crs as ccrs
from scipy import stats as sps

# ----- the only block that differs between the twin notebooks -----
A, B = 'CTRL_FUT', 'LCC_FUT'                # contrast = B − A
CASES = {A: 'NF2100ssp585norbc_tropstratchem_nudg_ctrl_f19_f19-20261028',
         B: 'NF2100ssp585norbc_tropstratchem_nudg_lcc_f19_f19-20261028'}
COL = {A: 'CTRL_FUT', B: 'LCC_FUT'}       # column names in the tables
DIFF = 'LCC − CTRL'
TAG = 'lcc_ctrl_fut'
OUT = Path('background_aerosol_lcc_ctrl_fut_online_outputs')
# -------------------------------------------------------------------

CASEDIR = Path.home() / 'cases/BRL_FRST_XPSN'
PROC = Path('../../data/processed-data/postprocessing')
PROC_YEARS = '2001-2025'
YEARS = list(range(2001, 2026))
YRS = f'{YEARS[0]}–{YEARS[-1]}'
DMS_DIR = Path('/cluster/shared/noresm/inputdata/noresm-only/atm/cam/camoslo')
MOD_FILE = Path('../lcc-modified-cells/modified_cells.nc')
CACHE, FIG, TAB = OUT / 'cache', OUT / 'figures', OUT / 'tables'  # cache = regenerable totals and maps
FIG_GLOBAL = FIG / 'appendix-global'
for d in (CACHE, FIG, TAB, FIG_GLOBAL):
    d.mkdir(parents=True, exist_ok=True)
REBUILD_CACHE = False
N_WORKERS = 8
R_EARTH = 6.371e6
DPM = np.array([31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31])  # noleap
LAT_MAP = 45   # Arctic maps: north of this latitude
LAT_BOX = 40   # regional totals (bars, time series): north of this latitude
ALPHA_FDR = 0.10
APPENDIX = []  # map calls, replayed globally in the Appendix
# project palette, in order; gold = total; first run = navy, second run = red
TEAL, NAVY, PLUM, RED, ORANGE, GOLD, GREY = '#5D9A92', '#2C3990', '#9A4A84', '#BC423E', '#DE7A3A', '#EBC45C', '#CCCCCC'
PCOL = {A: NAVY, B: RED}

def cell_area(lat, nlon):
    lat = np.asarray(lat, float)
    edges = np.clip(np.concatenate([[-90], 0.5 * (lat[1:] + lat[:-1]), [90]]), -90, 90)
    band = R_EARTH**2 * np.deg2rad(360 / nlon) * np.diff(np.sin(np.deg2rad(edges)))
    return np.repeat(band[:, None], nlon, axis=1)  # m2

with xr.open_dataset(PROC / f'{A}_atm_METEO_{PROC_YEARS}.nc') as g:
    AREA = xr.DataArray(cell_area(g.lat, g.lon.size), dims=('lat', 'lon'), coords=dict(lat=g.lat.values, lon=g.lon.values))
assert abs(AREA.sum() / (4 * np.pi * R_EARTH**2) - 1) < 1e-6
with xr.open_dataset(MOD_FILE) as m:
    assert np.allclose(m.lat, AREA.lat) and np.allclose(m.lon, AREA.lon)
    MOD = (m.modified.astype(bool)).assign_coords(lat=AREA.lat, lon=AREA.lon).load()
LAT = AREA.lat.broadcast_like(AREA)
REGIONS = {'global': AREA, 'NH': AREA * (LAT > 0), 'SH': AREA * (LAT < 0), '46–82°N': AREA * ((LAT >= 46) & (LAT <= 82)),
           'modified cells': AREA * MOD, 'Arctic >66.5°N': AREA * (LAT > 66.5)}
print(f'contrast: {B} − {A}')
print('modified cells:', int(MOD.sum()), f'({float((AREA * MOD).sum()) / 1e12:.2f} million km²), lat',
      float(LAT.where(MOD).min()), '–', float(LAT.where(MOD).max()))

# --- statistics (as src/boreal_forest_expansion/analysis/stats.py) ---
def pvalue(d, dim='year'):
    """Two-sided one-sample t-test of the paired yearly differences."""
    def _p(x):
        with warnings.catch_warnings(), np.errstate(invalid='ignore', divide='ignore'):
            warnings.simplefilter('ignore', RuntimeWarning)
            return sps.ttest_1samp(x, 0.0, axis=-1, nan_policy='omit').pvalue
    return xr.apply_ufunc(_p, d, input_core_dims=[[dim]])

def fdr_mask(p, within, alpha=ALPHA_FDR):
    """Benjamini-Hochberg over the cells in `within`."""
    pv = p.values.copy()
    valid = np.isfinite(pv) & within.values.astype(bool)
    sig = np.zeros_like(pv, dtype=bool)
    flat = pv[valid]
    if flat.size:
        ranked = np.sort(flat)
        below = np.nonzero(ranked <= alpha * np.arange(1, flat.size + 1) / flat.size)[0]
        if below.size:
            sig[valid] = flat <= ranked[below.max()]
    return xr.DataArray(sig, coords=p.coords, dims=p.dims)

def mean_ci_p(d):
    d = np.asarray(d, float)
    with warnings.catch_warnings():
        warnings.simplefilter('ignore', RuntimeWarning)
        p = sps.ttest_1samp(d, 0.0).pvalue
    return d.mean(), sps.t.ppf(0.975, len(d) - 1) * d.std(ddof=1) / np.sqrt(len(d)), p

def contrast_row(ya, yb):
    m, ci, p = mean_ci_p(yb - ya)
    return {COL[A]: ya.mean(), COL[B]: yb.mean(), DIFF: m, 'change %': 100 * m / ya.mean() if ya.mean() else np.nan, '95 % CI ±': ci, 'p': p}

def region_table(fields, fname):
    """fields: {name: (unit, yearly A, yearly B, factor to the table unit, 'mean'|'sum')}."""
    rows = []
    for name, (unit, a, b, fac, how) in fields.items():
        for r, w in REGIONS.items():
            agg = (lambda x: (x * w).sum(('lat', 'lon'))) if how == 'sum' else (lambda x: (x * w).sum(('lat', 'lon')) / w.sum())
            rows.append(dict(variable=name, unit=unit, region=r, **contrast_row(agg(a).values * fac, agg(b).values * fac)))
    tab = pd.DataFrame(rows).set_index(['variable', 'unit', 'region'])
    tab.to_csv(TAB / fname)
    return tab

def show(tab, regions=('global', 'modified cells', '46–82°N', 'Arctic >66.5°N')):
    with pd.option_context('display.float_format', '{:.3g}'.format, 'display.max_rows', 200, 'display.width', 200):
        for r in regions:
            print(r); display(tab.xs(r, level='region'))

# --- maps: A, B, B − A; hatching = not significant (as forcing_maps.add_stippling style='lines') ---
def nice_levels(vmax, decades=3):
    """1-2-5 levels from the first one >= vmax down by `decades` decades."""
    e = np.floor(np.log10(vmax))
    cand = np.array([m * 10.0**k for k in np.arange(e - decades - 1, e + 2) for m in (1, 2, 5)])
    hi = cand[cand >= vmax * 0.999][0]
    return cand[(cand <= hi) & (cand >= hi / 10**decades * 0.999)]

def cyclic(da):
    return xr.concat([da, da.isel(lon=0).assign_coords(lon=360.)], 'lon')

def map_axes(region, width=16):
    if region == 'global':
        return plt.subplots(1, 3, figsize=(width, 3.4), subplot_kw=dict(projection=ccrs.Robinson()))
    fig, axs = plt.subplots(1, 3, figsize=(width, 6.2), subplot_kw=dict(projection=ccrs.NorthPolarStereo()))
    th = np.linspace(0, 2 * np.pi, 200)
    for ax in axs:
        ax.set_extent([-180, 180, LAT_MAP, 90], ccrs.PlateCarree())
        ax.set_boundary(mpath.Path(np.c_[np.sin(th), np.cos(th)] * 0.5 + 0.5), transform=ax.transAxes)
    return fig, axs

def white_zero(cmap, levels, extend='both'):
    """Same colours per bin as BoundaryNorm(levels, cmap.N, extend), except the bin that contains 0 is pure white."""
    cm = plt.get_cmap(cmap)
    levels = np.asarray(levels, float)
    off, n_ext = {'neither': (0, 0), 'min': (1, 1), 'max': (0, 1), 'both': (1, 2)}[extend]
    nreg = len(levels) - 1 + n_ext
    cols = cm(((cm.N - 1) / (nreg - 1) * np.arange(nreg)).astype(int))
    for i in range(len(levels) - 1):
        if levels[i] < 0 < levels[i + 1]:
            cols[i + off] = (1, 1, 1, 1)
    return mcolors.ListedColormap(cols), mcolors.BoundaryNorm(levels, nreg, extend=extend)

def draw(ax, da, cmap, norm):
    d = cyclic(da)
    im = ax.pcolormesh(d.lon, d.lat, d, cmap=cmap, norm=norm, transform=ccrs.PlateCarree(), shading='nearest')
    ax.coastlines(lw=0.4)
    if isinstance(ax.projection, ccrs.NorthPolarStereo):
        ax.plot(np.linspace(0, 360, 361), np.full(361, 66.5), ls='--', lw=0.6, color='k', transform=ccrs.PlateCarree())
    return im

def hatch(ax, mask):
    da = cyclic(mask.astype(float))
    cs = ax.contourf(da.lon, da.lat, np.where(da > 0.5, 1.0, np.nan), levels=[0.5, 1.5], colors='none',
                     hatches=['/////'], transform=ccrs.PlateCarree(), zorder=4)
    cs.set_edgecolor('dimgrey'); cs.set_linewidth(0.0); cs.set_hatch_linewidth(0.3)

def outline_modified(ax):
    da = cyclic(MOD.astype(float))
    ax.contour(da.lon, da.lat, da, levels=[0.5], colors='k', linewidths=0.5, transform=ccrs.PlateCarree(), zorder=5)

def rmean(da, r):
    w = REGIONS[r]
    return float((da * w).sum() / w.sum())

def triptych(name, unit, a, b, dom, fname, levels=None, region='arctic'):
    """a, b: yearly fields. Absolute maps of A and B, and B − A with FDR hatching. Colour levels from the plotted domain."""
    d = b - a
    dm = d.mean('year')
    if not np.any(dm.values):
        print(f'{name}: identical in the two runs, no map'); return
    if region == 'arctic':
        APPENDIX.append((triptych, (name, unit, a, b, dom, fname), dict(levels=levels)))
    within = dom & (LAT >= LAT_MAP) if region == 'arctic' else dom
    am, bm = a.mean('year'), b.mean('year')
    if levels is None:
        vmax = float(np.nanpercentile(np.r_[am.where(within).values.ravel(), bm.where(within).values.ravel()], 99.5))
        levels = nice_levels(vmax, 3)
    cm = plt.get_cmap('YlOrBr').copy(); cm.set_under('white')
    norm = mcolors.BoundaryNorm(levels, cm.N, extend='max')
    dmax = float(np.nanpercentile(np.abs(dm.where(within).values), 99.5))
    v = dm.where(within).values
    v = v[np.isfinite(v) & (v != 0)]
    if max((v > 0).mean(), (v < 0).mean()) >= 0.9:  # one-signed field: linear levels, so the pattern stays visible
        e = 10 ** np.floor(np.log10(dmax / 6))
        dl = min(m * e for m in (1, 2, 2.5, 5, 10) if m * e >= dmax / 6 * 0.999) * np.arange(1, 7)
    else:                                           # mixed signs: 1-2-5 levels
        dl = nice_levels(dmax, 2)
    dlev = np.r_[-dl[::-1], dl]  # the central bin (-dl[0], dl[0]) is white
    fig, axs = map_axes(region)
    u = '' if unit == '1' else unit
    rl, rk = ('Arctic', 'Arctic >66.5°N') if region == 'arctic' else ('global', 'global')
    for ax, k, f in ((axs[0], A, am), (axs[1], B, bm)):
        im = draw(ax, f.where(dom), cm, norm)
        ax.set_title(f'{k}: {rl} {rmean(f, rk):.3g} {u}', fontsize=10)
    fig.colorbar(im, ax=axs[:2], orientation='horizontal', shrink=0.6, pad=0.04, ticks=levels, label=unit, format=lambda x, _: f'{x:g}')
    im = draw(axs[2], dm.where(dom), *white_zero('RdBu_r', dlev))
    hatch(axs[2], within & ~fdr_mask(pvalue(d), within))
    outline_modified(axs[2])
    base = rmean(am, rk)
    x = 100 * rmean(dm, rk) / base if base else np.nan
    pct = '' if not base else f' ({x:+.0f} %)' if abs(x) >= 10 else f' ({x:+.2g} %)'
    axs[2].set_title(f'{DIFF}: {rl} {rmean(dm, rk):+.3g}{pct}\nmodified cells {rmean(dm, "modified cells"):+.3g} {u}', fontsize=10)
    cb = fig.colorbar(im, ax=axs[2], orientation='horizontal', shrink=0.95, pad=0.04, ticks=dlev[::2] if len(dlev) > 12 else dlev,
                      label=unit, format=lambda x, _: f'{x:g}')
    cb.ax.tick_params(labelsize=7, rotation=45)
    where = f'Arctic, north of {LAT_MAP}°N (dashed = 66.5°N)' if region == 'arctic' else 'global'
    fig.suptitle(f'{name}, {YRS} mean, {where}\nhatched = not significant (paired t-test, FDR {ALPHA_FDR}); black line = modified cells', y=1.07, fontsize=11)
    fig.savefig((FIG if region == 'arctic' else FIG_GLOBAL) / fname, dpi=170, bbox_inches='tight')
    plt.show()
```

## 1. Inputs: what differs between the two runs

Resolved namelists (`CaseDocs/`) with the case names replaced by `<CASE>`; log-file names are ignored. At most 30 differing lines are printed per file, with the total count.

```python
def norm_lines(case, fname):
    txt = (CASEDIR / case / 'CaseDocs' / fname).read_text()
    return [l.rstrip() for l in txt.replace(case, '<CASE>').splitlines() if 'logfile' not in l]

files = sorted({f.name for c in CASES.values() for f in (CASEDIR / c / 'CaseDocs').iterdir() if f.is_file()})
changed = {}
for fname in files:
    dl = [x[:170] for x in difflib.unified_diff(norm_lines(CASES[A], fname), norm_lines(CASES[B], fname), A, B, n=0, lineterm='')
          if x[:1] in '+-' and x[:3] not in ('+++', '---')]
    if dl:
        changed[fname] = len(dl)
        print(f'=== {fname} ({len(dl)} differing lines)'); print('\n'.join(dl[:30]))
print('\nfiles compared:', len(files), '| with differences:', changed)
for key in ('SSTICE_DATA_FILENAME', 'RUN_REFCASE', 'RUN_REFDATE'):
    v = {k: re.search(rf'id="{key}" value="([^"]*)"', ''.join(f.read_text() for f in sorted((CASEDIR / c).glob('env_*.xml')))).group(1)
         for k, c in CASES.items()}
    print(f'{key}: ' + ('same: ' + v[A] if v[A] == v[B] else str(v)))
```

## 2. Online sources: MEGAN BVOC, dust, sea salt, ocean DMS flux, lightning NOx

| Source | What drives it in the model | How it is computed here | Evidence |
|---|---|---|---|
| **MEGAN BVOC** (19 CAM species) | CLM vegetation, leaf area, temperature, light, soil moisture, CO₂ | CLM `MEG_<compound>` × `landfrac`, summed into CAM species with `megan_specifier` | (a) model output; checked vs CAM `MEG_ISOP`, `MEG_MTERP` |
| **Dust** | CLM: wind, soil moisture, snow, vegetation cover | raw CLM `DSTFLXT` × `landfrac` × erodibility / `dust_emis_fact` × 1.15 | formula (a); absolute value (b): no CAM dust-emission output |
| **Sea salt** (3 modes, number flux) | 10-m wind (∝ U¹⁰^3.74), open ocean, SST | Salter et al. (2015) on **monthly means** of `U10`, `OCNFRAC`, `ICEFRAC` and prescribed SST | (b): underestimates the absolute flux; ratios are more robust |
| **Ocean DMS flux** | seawater DMS, transfer velocity (U10², SST), open ocean | `oslo_aero_dms_emis` (HAMOCC branch) on monthly means | (b), same caveat |
| **Lightning NOx** | convective cloud-top height (Price & Rind) | global value CAM prints every time step in `atm.log` | (a); global only, no map |

No CAM output of sea-salt, dust or DMS emission exists in these runs.

```python
import gzip
ARCHIVES = [Path('/cluster/work/users/adelez/archive'), Path.home() / 'BRL-FRST-XPSN_archive']  # live archive first (has 2015–2025)
SEC_YR = 365 * 86400
MW_DMS, MW_S = 62.13, 32.06
# Salter et al. (2015) sea-salt number flux, 3 modes: coefficients of SST³, SST², SST, 1 (oslo_aero_seasalt.F90)
SS_COEF = np.array([[-5.2168e5, 3.31725e7, -6.95275e8, 1.0684e10],
                    [0.0,       7.374e5,  -2.4803e7,   7.7373e8],
                    [0.0,       1.4210e4,  1.4662e7,   1.7075e8]])

def read_nml(case, key, fname='atm_in'):
    txt = (CASEDIR / case / 'CaseDocs' / fname).read_text()
    return re.search(rf"\b{key}\s*=\s*(.*?)\n\s*(?:\w+\s*=|/)", txt, re.S).group(1).strip()

def xml_value(case, key):
    for f in sorted((CASEDIR / case).glob('env_*.xml')):
        m = re.search(rf'id="{key}" value="([^"]*)"', f.read_text())
        if m:
            return m.group(1)
    raise KeyError(key)

def yearly(da, years):
    """Monthly means -> calendar-year means weighted by days per month; dims (year, lat, lon)."""
    w = da.time.dt.days_in_month
    out = (da * w).groupby('time.year').sum('time') / w.groupby('time.year').sum('time')
    assert set(years) <= set(out.year.values.tolist()), (years, out.year.values)
    return out.sel(year=list(years))

def hist_files(case, comp, years):
    """Raw monthly h0 files of `years`, from the first archive that has all of them."""
    tag = 'clm2' if comp == 'lnd' else 'cam'
    for root in ARCHIVES:
        fs = [root / case / comp / 'hist' / f'{case}.{tag}.h0.{y}-{m:02d}.nc' for y in years for m in range(1, 13)]
        if all(f.is_file() for f in fs):
            return fs
    raise FileNotFoundError(f'{case} {comp} h0 {years[0]}–{years[-1]}')

# --- MEGAN: CAM species <- CLM compounds (megan_specifier in drv_flds_in) ---
def megan_groups(case):
    spec = re.findall(r"'([^']*)'", read_nml(case, 'megan_specifier', 'drv_flds_in'))
    return {k.strip(): [c.strip() for c in v.split('+')] for k, v in (s.split('=') for s in spec)}

def megan_yearly(alias, case, years):
    """kg m-2 s-1 per grid-cell area: sum of the CLM compounds of each CAM species x landfrac."""
    with xr.open_dataset(PROC / f'{alias}_lnd_BVOC_{PROC_YEARS}.nc') as ds:
        assert ds.attrs['casename'] == case, (alias, ds.attrs['casename'])
        lf = ds.landfrac.fillna(0)
        lf = lf.isel(time=0) if 'time' in lf.dims else lf
        out = {sp: yearly(sum(ds[f'MEG_{c}'].fillna(0) for c in comps) * lf, years)
               for sp, comps in megan_groups(case).items()}
    return xr.Dataset(out).assign_coords(lat=AREA.lat, lon=AREA.lon).load()

# --- dust: CLM DSTFLXT x landfrac x erodibility / dust_emis_fact x 1.15 (oslo_aero_dust_emis) ---
def _dust_month(args):
    path, fac = args
    with netCDF4.Dataset(path) as d:
        return np.ma.filled(d['DSTFLXT'][0], 0.) * np.ma.filled(d['landfrac'][:], 0.) * fac

def dust_yearly(alias, case, years):
    f = CACHE / f'dust_{alias}_{years[0]}-{years[-1]}.nc'
    if f.is_file() and not REBUILD_CACHE:
        da = xr.open_dataarray(f).load()
        assert np.allclose(da.lat, AREA.lat) and np.allclose(da.lon, AREA.lon)
        return da.assign_coords(lat=AREA.lat, lon=AREA.lon)  # same grid; avoid an outer join on float coords
    erod = xr.open_dataset(read_nml(case, 'soil_erod_file').strip("'")).mbl_bsn_fct_geo
    assert erod.shape == AREA.shape and np.allclose(erod.lat[1:-1], AREA.lat[1:-1])  # same grid (file poles at ±89.5)
    fac = np.where(erod.values < 0.1, 0., erod.values) / float(read_nml(case, 'dust_emis_fact').replace('D0', '')) * 1.15
    files = hist_files(case, 'lnd', years)
    with ProcessPoolExecutor(N_WORKERS) as ex:
        mon = np.stack(list(ex.map(_dust_month, [(p, fac) for p in files]))).reshape(len(years), 12, *AREA.shape)
    da = xr.DataArray((mon * DPM[None, :, None, None]).sum(1) / DPM.sum(), dims=('year', 'lat', 'lon'),
                      coords=dict(year=list(years), lat=AREA.lat, lon=AREA.lon), name='dust',
                      attrs=dict(units='kg m-2 s-1', case=case))
    da.to_netcdf(f)
    return da

# --- ocean: DMS flux and sea-salt number flux, offline from monthly means with the model formulas ---
def ocean_yearly(alias, case, years):
    """oslo_aero_ocean.F90 (DMS 'lana', HAMOCC transfer velocity) and oslo_aero_seasalt.F90 (Salter et al. 2015).
    Inputs: model U10, OCNFRAC, ICEFRAC (monthly), prescribed SST and seawater DMS (monthly climatologies)."""
    with xr.open_dataset(PROC / f'{alias}_atm_METEO_{PROC_YEARS}.nc') as ds:
        assert ds.attrs['casename'] == case, (alias, ds.attrs['casename'])
        m = ds[['U10', 'OCNFRAC', 'ICEFRAC']].load()
    with xr.open_dataset(xml_value(case, 'SSTICE_DATA_FILENAME'), decode_times=False) as s:
        sst = s.SST_cpl_prediddle.load()   # deg C
    with xr.open_dataset(DMS_DIR / read_nml(case, 'ocean_filename').strip("'"), decode_times=False) as s:
        dms = s.dms.fillna(0).load()       # nmol L-1
    for x in (sst, dms):
        assert x.sizes['time'] == 12 and np.allclose(x.lat, m.lat) and np.allclose(x.lon, m.lon)
    idx = m.time.dt.month.values - 1
    like = lambda x: xr.DataArray(x.values[idx], dims=m.U10.dims, coords=m.U10.coords)
    t, c, u = like(sst), like(dms), m.U10
    open_ = (m.OCNFRAC * (1 - m.ICEFRAC)).clip(0, 1)
    sc = 2855.7 + (-177.63 + (6.0438 + (-0.11645 + 0.00094743 * t) * t) * t) * t   # Schmidt number of DMS
    fdms = (MW_DMS * 6.97e-7 * open_ * u**2 * np.sqrt(660 / sc) * 1e-9 * c).fillna(0)  # kg DMS m-2 s-1
    wc = 2e-8 * u**3.74 * open_                                                        # whitecap fraction
    out = dict(DMS_flux=yearly(fdms, years), open_ocean=yearly(open_, years))
    for n, (a, b, cc, d) in enumerate(SS_COEF, 1):
        out[f'seasalt_n{n}'] = yearly((wc * (((a * t + b) * t + cc) * t + d)).fillna(0).clip(min=0), years)  # m-2 s-1
    return xr.Dataset(out).assign_coords(lat=AREA.lat, lon=AREA.lon).load()

# --- lightning NOx: CAM prints the global value every time step in atm.log ---
def lightning_yearly(case, years):
    f = CACHE / f'lightning_{case}.csv'
    if f.is_file() and not REBUILD_CACHE:
        s = pd.read_csv(f, index_col=0).iloc[:, 0]
    else:
        steps = int(xml_value(case, 'ATM_NCPL')) * 365
        logs = {}
        for root in ARCHIVES:
            for g in sorted((root / case / 'logs').glob('atm.log.*.gz')):
                logs.setdefault(g.name, g)
        out = {}
        for g in logs.values():
            y0, v = None, []
            with gzip.open(g, 'rt', errors='replace') as fh:
                for line in fh:
                    if 'lightning NOx (TgN/y)' in line:
                        v.append(float(line.rsplit('=', 1)[1].split()[1]))
                    elif y0 is None and 'Current date (yr mon day tod)' in line:
                        y0 = int(line.split(':')[1].split()[0])
            if len(v) % steps == 1:   # the first segment also prints step 0
                v = v[1:]
            assert len(v) % steps == 0, (g.name, len(v))
            for k in range(len(v) // steps):
                assert y0 + k not in out, (g.name, y0 + k)
                out[y0 + k] = float(np.mean(v[k * steps:(k + 1) * steps]))
        s = pd.Series(out, name='lightning NOx (Tg N/yr)').sort_index()
        s.to_csv(f)
    missing = set(years) - set(s.index)
    assert not missing, (case, sorted(missing))
    return s.loc[list(years)]
```

```python
ONLINE = {  # panel -> {component: (function of SRC[run] -> kg or # m-2 s-1, factor to the panel unit)}
    'MEGAN BVOC (Tg)': {'isoprene': (lambda d: d.ISOP, 1e-9), 'monoterpenes': (lambda d: d.MTERP, 1e-9),
                        'sesquiterpenes': (lambda d: d.BCARY, 1e-9), 'methanol': (lambda d: d.CH3OH, 1e-9),
                        'other 15 species': (lambda d: sum(d[v] for v in MEG_OTHER), 1e-9)},
    'dust (Tg)': {'dust': (lambda d: d.dust, 1e-9)},
    'DMS flux, offline (Tg S)': {'DMS flux': (lambda d: d.DMS_flux * MW_S / MW_DMS, 1e-9)},
    'sea salt number, offline (10²⁷)': {f'sea salt mode {n}': (lambda d, n=n: d[f'seasalt_n{n}'], 1e-27) for n in (1, 2, 3)},
}
def online_totals(w):
    return {(k, panel): pd.DataFrame({c: (fn(SRC[k]) * w).sum(('lat', 'lon')).to_series() * SEC_YR * fac
                                      for c, (fn, fac) in comps.items()})
            for k in CASES for panel, comps in ONLINE.items()}
ONL = online_totals(AREA)
ONL_N = online_totals(AREA.where(AREA.lat >= LAT_BOX, 0))
for k in CASES:
    ONL[k, 'lightning NOx (Tg N)'] = pd.DataFrame({'lightning NOx': LIGHT[k]})
PANELS, PANELS_N = list(ONLINE) + ['lightning NOx (Tg N)'], list(ONLINE)
COMPCOL = dict(zip(['isoprene', 'monoterpenes', 'sesquiterpenes', 'methanol', 'other 15 species'], [TEAL, NAVY, PLUM, RED, ORANGE]))
COMPCOL.update({'sea salt mode 1': TEAL, 'sea salt mode 2': NAVY, 'sea salt mode 3': PLUM, 'dust': TEAL, 'DMS flux': TEAL, 'lightning NOx': TEAL, 'total': GOLD})
assert all(list(ONL[k, p].index) == YEARS for k in CASES for p in PANELS)
pd.concat({f'{k}|{p}': ONL[k, p] for k in CASES for p in PANELS}, names=['run|panel', 'year']).to_csv(TAB / f'online_sources_global_yearly_{PROC_YEARS}.csv')
pd.concat({f'{k}|{p}': ONL_N[k, p] for k in CASES for p in PANELS_N}, names=['run|panel', 'year']).to_csv(TAB / f'online_sources_north{LAT_BOX}N_yearly_{PROC_YEARS}.csv')
assert np.isclose(ONL[A, 'MEGAN BVOC (Tg)'].isoprene.mean(), SRC_TAB.loc[('MEGAN isoprene', 'Tg yr⁻¹', 'global'), COL[A]], rtol=1e-6)
assert np.isclose(ONL[B, 'dust (Tg)'].dust.mean(), SRC_TAB.loc[('dust', 'Tg yr⁻¹', 'global'), COL[B]], rtol=1e-6)

def online_bars(ONL, PANELS, where, fname):
    fig, axs = plt.subplots(1, len(PANELS), figsize=(17, 3.8), gridspec_kw=dict(wspace=0.55))
    for ax, panel in zip(axs, PANELS):
        bottom = np.zeros(2)
        for c in ONL[A, panel].columns:
            v = np.array([ONL[k, panel][c].mean() for k in CASES])
            ax.bar([0, 1], v, bottom=bottom, color=COMPCOL[c], edgecolor='white', lw=0.5, label=c)
            bottom += v
        sd = [ONL[k, panel].sum(axis=1).std(ddof=1) for k in CASES]
        ax.errorbar([0, 1], bottom, yerr=sd, fmt='none', ecolor='k', elinewidth=0.8, capsize=3)
        for x, (v, e) in enumerate(zip(bottom, sd)):
            ax.text(x, v + e, f'{float(f"{v:.3g}"):g}', ha='center', va='bottom', fontsize=8)
        ax.set_xticks([0, 1], list(CASES), fontsize=8)
        ax.set_title(panel, fontsize=10); ax.set_ylim(0, (bottom + sd).max() * 1.15)
        ax.spines[['top', 'right']].set_visible(False)
    h, l = [], []
    for ax in axs:
        for hh, ll in zip(*ax.get_legend_handles_labels()):
            if ll not in l and ll not in ('dust', 'DMS flux', 'lightning NOx'):
                h.append(hh); l.append(ll)
    fig.legend(h, l, loc='upper center', ncol=4, frameon=False, bbox_to_anchor=(0.5, -0.06))
    fig.suptitle(f'Online sources: {where} total per year, mean {YRS} (error bar = ±1 s.d. of the yearly totals)', y=1.02)
    fig.savefig(FIG / fname, dpi=200, bbox_inches='tight'); plt.show()

def effect_bars(ONL, PANELS, where, fname):
    ratios = [max(ONL[A, p].shape[1] + (ONL[A, p].shape[1] > 1), 1.3) for p in PANELS]
    fig, axs = plt.subplots(1, len(PANELS), figsize=(17, 3.8), gridspec_kw=dict(wspace=0.42, width_ratios=ratios))
    for ax, panel in zip(axs, PANELS):
        d = ONL[B, panel] - ONL[A, panel]
        if d.shape[1] > 1:
            d = d.assign(total=d.sum(axis=1))
        s = [mean_ci_p(d[c].values) for c in d.columns]
        ax.bar(range(d.shape[1]), [x[0] for x in s], 0.7, yerr=[x[1] for x in s], color=[COMPCOL[c] for c in d.columns],
               error_kw=dict(elinewidth=0.8, capsize=2.5))
        ax.axhline(0, color='k', lw=0.6)
        ax.set_xticks(range(d.shape[1]), [c.replace(' species', '').replace('sea salt ', '').replace('sesquiterpenes', 'sesquit.').replace('monoterpenes', 'monot.') for c in d.columns],
                      rotation=40, ha='right', fontsize=8)
        ax.set_title(panel, fontsize=10); ax.spines[['top', 'right']].set_visible(False)
    fig.suptitle(f'{B} − {A}: {where}, mean {YRS} (error bar = 95 % CI of the paired yearly differences)', y=1.02)
    fig.savefig(FIG / fname, dpi=200, bbox_inches='tight'); plt.show()

def series(ONL, PANELS, where, fname, difference=False):
    fig, axs = plt.subplots(2, 3, figsize=(17, 7.5))
    for ax in axs.flat[len(PANELS):-1]:
        ax.axis('off')
    for ax, panel in zip(axs.flat, PANELS):
        todo = [(ONL[B, panel] - ONL[A, panel], '-', None, True)] if difference else [(ONL[A, panel], '-', None, True), (ONL[B, panel], '--', 'o', False)]
        for s, ls, mk, lab in todo:
            multi = s.shape[1] > 1
            for c in s.columns:
                ax.plot(s.index, s[c].values, ls=ls, marker=mk, ms=2.5, lw=1, color=COMPCOL[c] if multi else GOLD,
                        label=(c if multi else 'total') if lab else None)
            if multi:
                ax.plot(s.index, s.sum(axis=1).values, ls=ls, marker=mk, ms=3, lw=2.2, color=GOLD, label='total (sum of components)' if lab else None)
        if difference:
            ax.axhline(0, color='k', lw=0.6)
        else:
            ax.set_ylim(bottom=0)
        ax.set_title(('Δ ' if difference else '') + panel, fontsize=10)
        ax.set_xlim(YEARS[0] - 0.5, YEARS[-1] + 0.5); ax.set_xticks(range(2001, 2026, 4))
        ax.spines[['top', 'right']].set_visible(False)
    h, l = [], []
    for ax in axs.flat[:len(PANELS)]:
        for hh, ll in zip(*ax.get_legend_handles_labels()):
            if ll not in l and ll != 'total':
                h.append(hh); l.append(ll)
    if not difference:
        h += [plt.Line2D([], [], color='k', ls='-'), plt.Line2D([], [], color='k', ls='--', marker='o', ms=3)]
        l += [A, B]
    axs.flat[-1].axis('off'); axs.flat[-1].legend(h, l, loc='center', frameon=False, fontsize=9)
    what = f'{B} − {A}' if difference else f'solid = {A}, dashed = {B}'
    fig.suptitle(f'Online sources: {where} total of each year, {YRS}; {what} (single-component panels: gold)')
    fig.tight_layout(); fig.savefig(FIG / fname, dpi=200, bbox_inches='tight'); plt.show()

for O, P, where, tag in ((ONL, PANELS, 'global', ''), (ONL_N, PANELS_N, f'north of {LAT_BOX}°N', f'_north{LAT_BOX}N')):
    online_bars(O, P, where, f'online_totals_mean{tag}.png')
    effect_bars(O, P, where, f'online_difference_mean{tag}.png')
    series(O, P, where, f'online_timeseries{tag}.png')
    series(O, P, where, f'online_difference_timeseries{tag}.png', difference=True)
```

## 3. What the model makes of it

Processed monthly files, yearly means weighted by days per month; surface = lowest model level. SOA burden = `cb_SOA_A1 + cb_SOA_NA + cb_SOA_A1_OCW + cb_SOA_NA_OCW` (interstitial + cloud-borne). A variable that is identical in the two runs gets a table row but no map.

```python
MODEL_VARS = {  # name: (category, variable(s), surface?, factor, unit)
    'AOD 550 nm': ('AEROPT', ['DOD550'], False, 1, '1'),
    'absorption AOD 550 nm': ('AEROPT', ['ABS550'], False, 1, '1'),
    'SOA burden': ('SOA', ['cb_SOA_A1', 'cb_SOA_NA', 'cb_SOA_A1_OCW', 'cb_SOA_NA_OCW'], False, 1e6, 'mg m⁻²'),
    'sulfate burden': ('SULFATE', ['cb_SULFATE'], False, 1e6, 'mg m⁻²'),
    'droplet column': ('CLOUDPROP', ['CDNUMC'], False, 1e-10, '10¹⁰ m⁻²'),
    'surface isoprene': ('BVOC', ['ISOP'], True, 1e12, 'ppt'),
    'surface monoterpenes': ('BVOC', ['MTERP'], True, 1e12, 'ppt'),
    'surface OH': ('CHEM', ['OH'], True, 1e15, 'ppq'),
    'surface SO₂': ('SULFATE', ['SO2'], True, 1e12, 'ppt'),
    'surface DMS': ('SULFATE', ['DMS'], True, 1e12, 'ppt'),
    'surface NOx': ('CHEM', ['NO', 'NO2'], True, 1e12, 'ppt'),
    'surface O₃': ('CHEM', ['O3'], True, 1e9, 'ppb'),
    'surface CO': ('CHEM', ['CO'], True, 1e9, 'ppb'),
    'surface CH₄': ('CHEM', ['CH4'], True, 1e9, 'ppb'),
    'O₃ column': ('CHEM', ['cb_O3'], False, 1e3, 'g m⁻²'),
}

def load_model(k):
    out, files = {}, {}
    for name, (cat, vs, sfc, fac, unit) in MODEL_VARS.items():
        f = PROC / f'{k}_atm_{cat}_{PROC_YEARS}.nc'
        if f not in files:
            files[f] = xr.open_dataset(f)
            assert files[f].attrs['casename'] == CASES[k], (f, files[f].attrs['casename'])
            assert files[f].time.encoding.get('calendar') == 'noleap'
        ds = files[f]
        if sfc:
            assert float(ds.lev[-1]) > 900
        da = sum(ds[v].isel(lev=-1) if sfc else ds[v] for v in vs)
        assert np.allclose(da.lat, AREA.lat) and np.allclose(da.lon, AREA.lon)
        out[name] = (yearly(da, YEARS) * fac).assign_coords(lat=AREA.lat, lon=AREA.lon).load()
    for ds in files.values():
        ds.close()
    return out

MODEL = {k: load_model(k) for k in CASES}
MODEL_TAB = region_table({n: (MODEL_VARS[n][4], MODEL[A][n], MODEL[B][n], 1, 'mean') for n in MODEL_VARS},
                         f'model_levels_{TAG}_{PROC_YEARS}.csv')
show(MODEL_TAB)
```

### 3.1 Far-field check: tropical isoprene (LCC only)

The land cover differs only at 46–82°N, yet the global maps show isoprene differences in the Amazon and Congo. This cell tests whether they are a persistent response or internal variability plus a start-up transient: tropical (30°S–30°N) MEGAN isoprene by period, the gross cell-by-cell difference, the correlation between the difference patterns of two independent decades, and the share of tropical land cells with uncorrected p < 0.05 (5 % expected by chance).

```python
TROP, TG = (AREA.lat >= -30) & (AREA.lat <= 30), 1e-9 * SEC_YR
d = SRC[B].ISOP - SRC[A].ISOP
lnd = SRC[A].ISOP.mean('year') > 1e-13
net = (d * AREA).where(TROP).sum(('lat', 'lon')) * TG
gross = lambda x: float((abs(x) * AREA).where(TROP & lnd).sum()) * TG
d1, d2 = d.sel(year=slice(2006, 2015)).mean('year'), d.sel(year=slice(2016, 2025)).mean('year')
x = np.c_[d1.where(TROP & lnd).values.ravel(), d2.where(TROP & lnd).values.ravel()]
x = x[np.isfinite(x).all(1)]
pv = pvalue(d).where(TROP & lnd).values
FAR = pd.Series({'CTRL tropical isoprene (Tg/yr)': float((SRC[A].ISOP.mean('year') * AREA).where(TROP).sum()) * TG,
                 'net Δ 2001–2025': float(net.mean()), 'net Δ 2001': float(net.sel(year=2001)),
                 'net Δ 2002–2005': float(net.sel(year=slice(2002, 2005)).mean()),
                 'net Δ 2006–2025': float(net.sel(year=slice(2006, 2025)).mean()),
                 'gross |Δ| of cell means, 2006–2015': gross(d1), 'gross |Δ| of cell means, 2016–2025': gross(d2),
                 'pattern r, 2006–15 vs 2016–25': np.corrcoef(x.T)[0, 1],
                 '% tropical land cells p < 0.05': 100 * np.mean(pv[np.isfinite(pv)] < 0.05)}, name=DIFF)
FAR.to_csv(TAB / f'farfield_tropical_isoprene_{PROC_YEARS}.csv')
with pd.option_context('display.float_format', '{:.3g}'.format):
    display(FAR.to_frame())
```

Inferences (b):
- The forest expansion **does not change the FUT aerosol background** in a meaningful way: the sources that matter for aerosol (sulfur, fires, dust, sea salt) are unchanged, and the BVOC change is a **shift in composition** (more isoprene, fewer monoterpenes, sesquiterpenes and oxygenated VOCs) rather than more BVOC, so SOA barely moves.
- Monoterpenes, the main SOA precursor in this mechanism (yield 0.15 vs 0.025 for isoprene), go **down** although the forest expands. Why (PFT mix of the LPJ-GUESS land cover, emission factors, leaf temperature) is not checked here; the MEGAN LMDI split addresses it.
- The droplet-column decrease, with AOD/SOA/sulfate unchanged, points to the **meteorology** (cloud amount, boundary layer over a darker, rougher surface) rather than to aerosol. Not checked here.
- The O₃ increase and CO/NOx decrease fit more isoprene oxidation and a warmer, more reactive boundary layer. Not separated here.

**Added with the merged analysis** (a)

- **Arctic >66.5°N:** MEGAN monoterpene emissions rise by 28 % and sesquiterpenes by 28 %, from a small base, while they fall further south. The SOA burden rises by 0.54 % (p < 0.05, uncorrected), and surface isoprene falls by 4.3 %.
- **Tropical differences (§3.1):** tropical isoprene differs although the land cover does not change there. The net difference is +7.3 Tg in 2001 and +0.35 Tg yr⁻¹ in 2006–2025, and the cell pattern does not persist between 2006–15 and 2016–25 (r = 0.00). This looks like a start-up transient plus internal variability (b), so the formally significant global isoprene change (+0.15 %, p = 0.02) should not be read as a forced far-field response.
